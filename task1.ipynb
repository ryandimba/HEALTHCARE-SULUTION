```python
name = input('Enter my name: ')
year = int(input('Enter my year of birth: '))

if year <= 1980:
    print(f'{name}, I am an ancestor')
elif year <= 1995:
    print(f'{name}, I am a millennial')
else:
    print(f'{name}, I am a Gen Z')
```
